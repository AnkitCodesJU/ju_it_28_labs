# Water Jug Problem using Breadth-First Search (BFS)

**Problem Statement:**
You are given two jugs with capacities `Jug 1` and `Jug 2`, and neither has any measuring markers. You are also given a `Target` measurement. You must use these two jugs to measure exactly the target amount of water.

**Allowed Operations (Production Rules):**
1. **Fill** a jug completely.
2. **Empty** a jug completely on the ground.
3. **Pour** water from one jug to the other until the receiving jug is full or the pouring jug is empty.

**Why BFS?**
Breadth-First Search guarantees the **shortest path** to the solution. It explores all possible moves level by level, ensuring that the first time we reach the target state, it is achieved in the absolute minimum number of steps.

```python
import math
from collections import deque
import os

def solve_water_jug_bfs(jug1_cap, jug2_cap, target, output_file="output.txt"):
    """
    Solves the Water Jug problem using BFS and writes the intermediate steps to a file.
    """
    # 1. Edge Case Checks
    if target > max(jug1_cap, jug2_cap):
        print(f"Error: Target ({target}) cannot be greater than the maximum capacity ({max(jug1_cap, jug2_cap)}).")
        return

    # Mathematical rule: Target must be a multiple of the Greatest Common Divisor of the two capacities
    if target % math.gcd(jug1_cap, jug2_cap) != 0:
        print("Error: No solution is mathematically possible for these capacities and target.")
        return

    # 2. Initialize BFS Queue and Visited Set
    # queue stores tuples of: ((current_jug1, current_jug2), path_of_actions)
    queue = deque()
    start_state = (0, 0)
    queue.append((start_state, [(0, 0, "Initial state")]))

    visited = set()
    visited.add(start_state)
    solution_path = None

    # 3. BFS Loop
    while queue:
        (curr_x, curr_y), path = queue.popleft()

        # Check if we reached the goal state
        if curr_x == target or curr_y == target:
            solution_path = path
            break

        # Generate all possible next moves
        moves = []

        # Rule 1: Fill Jug 1
        if curr_x < jug1_cap:
            moves.append((jug1_cap, curr_y, "Fill Jug 1 completely"))

        # Rule 2: Fill Jug 2
        if curr_y < jug2_cap:
            moves.append((curr_x, jug2_cap, "Fill Jug 2 completely"))

        # Rule 3: Empty Jug 1
        if curr_x > 0:
            moves.append((0, curr_y, "Empty Jug 1 on the ground"))

        # Rule 4: Empty Jug 2
        if curr_y > 0:
            moves.append((curr_x, 0, "Empty Jug 2 on the ground"))

        # Rule 5: Pour Jug 2 -> Jug 1
        if curr_y > 0 and curr_x < jug1_cap:
            amount_to_pour = min(curr_y, jug1_cap - curr_x)
            moves.append((curr_x + amount_to_pour, curr_y - amount_to_pour, "Pour water from Jug 2 to Jug 1"))

        # Rule 6: Pour Jug 1 -> Jug 2
        if curr_x > 0 and curr_y < jug2_cap:
            amount_to_pour = min(curr_x, jug2_cap - curr_y)
            moves.append((curr_x - amount_to_pour, curr_y + amount_to_pour, "Pour water from Jug 1 to Jug 2"))

        # Validate and add new states to queue
        for next_x, next_y, rule_used in moves:
            next_state = (next_x, next_y)

            if next_state not in visited:
                visited.add(next_state)
                # Copy current path and append the new move
                new_path = list(path)
                new_path.append((next_x, next_y, rule_used))
                queue.append((next_state, new_path))

    # 4. Save to Output File & Print
    if solution_path:
        with open(output_file, 'w') as f:
            header = f"{'S.No.':<6} | {'Jug 1':<6} | {'Jug 2':<6} | {'Rule followed'}"
            f.write(header + "\n" + "-" * 65 + "\n")
            print(header + "\n" + "-" * 65)

            for idx, (x, y, rule) in enumerate(solution_path):
                line = f"{idx+1:<6} | {x:<6} | {y:<6} | {rule}"
                f.write(line + "\n")
                print(line)

        print(f"\n [Success] Intermediate outputs saved in Colab to: {os.path.abspath(output_file)}")
    else:
        print("No solution found.")
```

```python
# Interactive Execution Block
try:
    print("--- Water Jug Problem Solver ---")
    j1 = int(input("Enter the capacity of Jug 1 (e.g., 4): "))
    j2 = int(input("Enter the capacity of Jug 2 (e.g., 3): "))
    t = int(input("Enter the target measurement (e.g., 2): "))

    print("\nCalculating path...\n")
    solve_water_jug_bfs(j1, j2, t, output_file="output.txt")
except ValueError:
    print("Invalid input. Please enter whole numbers only.")
```
